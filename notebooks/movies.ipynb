```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    words = text.split()
    words = [word for word in words if word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)
```

```python
data['tags'] = data['tags'].apply(preprocess_text)
```

```python
data['tags'][1]
```

```python
data = data.reset_index(drop=True)
```

```python
indices = pd.Series(data.index, index=data['title']).drop_duplicates()
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

```

```python
tfidf = TfidfVectorizer(max_features=50000, ngram_range=(1,2))
```

```python
tfidf_metrix = tfidf.fit_transform(data['tags'])
```

```python
tfidf_metrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n=10):
    if title not in indices:
        return ['Movie not found']
    idx = indices[title]
    sim_score = cosine_similarity(tfidf_metrix[idx], tfidf_metrix).flatten()
    similar_idx = sim_score.argsort()[::-1][1:n+1]
    return data['title'].iloc[similar_idx]
```

```python
recommend("Toy Story")
```

```python
import pickle

pickle.dump(tfidf_metrix, open('tfidf_metrix.pkl','wb'))
pickle.dump(indices, open('indices.pkl', 'wb'))
pickle.dump(tfidf, open('tfidf.pkl', 'wb'))
data.to_pickle('data.pkl')
```
